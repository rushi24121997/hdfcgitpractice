```python

from pyspark.sql.functions import *

data = [
    ("2025-01-01", "2025-01-10"),
    ("2025-02-15", "2025-03-01"),
    ("2025-05-20", "2025-05-25")
]

df = spark.createDataFrame(data, ["start_date", "end_date"])

df = df.withColumn("start_date", to_date(col("start_date"))) \
       .withColumn("end_date", to_date(col("end_date")))

df.show()
```

```python
df = df.withColumn('dateDiff',datediff(col('end_date'),col('start_date')))
df.show()
```

```python
df.withColumn("timestamp",current_timestamp().alias('now')).show()
```

```python
df.select(
    "start_date",
    dayofweek("start_date").alias("minus_5_days")
).show()
```
