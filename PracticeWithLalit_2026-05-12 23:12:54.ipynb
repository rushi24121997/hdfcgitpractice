```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("test").getOrCreate()
```

```python
df = spark.read.format("csv")\
    .option('header',True)\
    .option('inferSchema',True)\
    .option('mode','PERMISSIVE')\
    .option("sep",",")\
    .load('/Volumes/online10/default/externaldata/employee_data.csv')
```

```python
df.show()
```

```python
#1 Get employees whose salary > 50,000 and city = "Pune"
df.filter((df.salary>5000) & (df.city =='Pune')).show()
```

```python
# 2. Add Bonus Column every emp get 10% of bonus according to salary
df.withColumn('bonus',df.salary*0.1).show()
```

```python
# 3 Replace null salary with 0
test = df.filter(df.name == 'Emp_34')
test.fillna(0,['salary']).show()

#
test.fillna({'salary':0}).show()
```

```python
# 4. Rename emp_id → id, name → emp_name
df.withColumnsRenamed({'emp_id':'id',"name":"emp_name"}).show()
```

```python
# Remove duplicate employees
df.dropDuplicates(["department"]).show()
```

```python
# 6. Count employees in each city and give me city where count is greater than 10
from pyspark.sql.functions import col,expr
df.groupBy('city').count().filter(col('count')>10).show()

```

```python
# 7. Sort Employees by Salary Desc
# Problem: Highest salary first
df.orderBy(col('salary').desc()).show()
```

```python
# 8. Get Top 3 Salaries
#Problem: Fetch top 3 highest paid employees also give nth highest salary
from pyspark.sql.functions import col
df.orderBy(col('salary').desc()).limit(3).show()
df.orderBy(col('salary').desc()).limit(3).offset(2).show()
# offset will skip first 2 rows
```

```python
#9. Add Current Timestamp
#Problem: Add column load_time
from pyspark.sql.functions import current_timestamp,from_utc_timestamp

df.withColumn("load_time",current_timestamp()).show()

```

```python
#10. to convert utc into ist
df.withColumn('ist_time',from_utc_timestamp(current_timestamp(),'IST')).show()
```

```python
#11. Select Specific Columns
#Problem: Only show name & salary
df.select('name','salary').show()
```

```python
#12. Create Column Based on Condition
#Problem: If salary > 50k → "High" else "Low"
from pyspark.sql.functions import when
df.withColumn('Status',when(col('salary')>50000,'High').otherwise('Low')).show()
```

```python
#13. Filter Null Values
#Problem: Get rows where salary is null
df.filter(col('salary').isNull()).show()
```

```python
#16. Read CSV with Schema
#Problem: Read file with defined schema
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,FloatType,DateType

schema = StructType([
    StructField('id',IntegerType(),True),
    StructField('name',StringType(),True),
    StructField('salary',FloatType(),True),
    StructField('age',IntegerType(),True),
    StructField('department',StringType(),True),
    StructField('manager_id',IntegerType(),True),
    StructField('city',StringType(),True),
    StructField('gender',StringType(),True),
    StructField('joining_date',DateType(),True)
])
df = spark.read.csv("/Volumes/online10/default/externaldata/employee_data.csv",header=True,schema=schema)
df.printSchema()
```

```python
#17. Handle Corrupt Records
#Problem: Store corrupt rows separately
from pyspark.sql.functions import col
df1 = spark.read.format('csv')\
    .option('header','true')\
        .option('mode','DROPMALFORMED')\
            .option('sep',',')\
                .load('/Volumes/online10/default/externaldata/corruptdata.csv')
df2 = spark.read.format('csv')\
    .option('header','true')\
        .option('mode','PERMISSIVE')\
            .option('sep',',')\
                .load('/Volumes/online10/default/externaldata/corruptdata.csv')

df2.subtract(df1).show()
```

```python
#18. Add Multiple Columns
#Problem: Add tax and net_salary
df.withColumn('tax',col('salary')*0.8)\
    .withColumn('net_salary',col("salary")-col('salary')*0.8).show()
```

```python
#19. Filter Using IN Condition
#Problem: Get employees from Pune, Mumbai
from pyspark.sql.functions import col,lower,upper
df.filter(lower(col('city')).isin(['pune','mumbai'])).show()
```

```python
#20. Remove Rows with Any Null
#Problem: Drop rows having null in any column
df.dropna(how='any').show()
```
